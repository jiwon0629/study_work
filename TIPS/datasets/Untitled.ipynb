```python
import json
import os
import torch
import numpy as np
from typing import List, Tuple

def iou(boxA: torch.Tensor, boxB: torch.Tensor) -> torch.Tensor:
    xA = torch.max(boxA[:, None, 0], boxB[None, :, 0])
    yA = torch.max(boxA[:, None, 1], boxB[None, :, 1])
    xB = torch.min(boxA[:, None, 2], boxB[None, :, 2])
    yB = torch.min(boxA[:, None, 3], boxB[None, :, 3])

    interArea = torch.clamp(xB - xA, min=0) * torch.clamp(yB - yA, min=0)
    boxAArea = (boxA[:, 2] - boxA[:, 0]) * (boxA[:, 3] - boxA[:, 1])
    boxBArea = (boxB[:, 2] - boxB[:, 0]) * (boxB[:, 3] - boxB[:, 1])

    iou = interArea / (boxAArea[:, None] + boxBArea[None, :] - interArea)
    return iou

def mAP_calculate(
    answer_list: List[Tuple[int, List[int]]],
    result_data: List[List[float]],
    iou_threshold: float = 0.5,
) -> float:
    result_class = {}
    for pred in result_data:
        class_id, score, x_min, y_min, x_max, y_max = pred
        if class_id not in result_class:
            result_class[class_id] = []
        result_class[class_id].append((score, [x_min, y_min, x_max, y_max]))

    answer_class = {}
    for gt in answer_list:
        class_id, box = gt
        if class_id not in answer_class:
            answer_class[class_id] = []
        answer_class[class_id].append(box)

    average_precisions = []
    
    for class_id in answer_class:
        gt_boxes = torch.tensor(answer_class[class_id], dtype=torch.float32)
        pred_boxes = result_class.get(class_id, [])
        pred_boxes = sorted(pred_boxes, key=lambda x: x[0], reverse=True)

        if not pred_boxes:
            continue

        pred_boxes = torch.tensor([b[1] for b in pred_boxes], dtype=torch.float32)

        ious = iou(pred_boxes, gt_boxes)

        tp = torch.zeros(pred_boxes.shape[0])
        fp = torch.zeros(pred_boxes.shape[0])
        matched_gt_boxes = set()

        for i in range(pred_boxes.shape[0]):
            iou_vals = ious[i]
            max_iou, max_iou_idx = torch.max(iou_vals, dim=0)

            if max_iou >= iou_threshold and max_iou_idx.item() not in matched_gt_boxes:
                tp[i] = 1
                matched_gt_boxes.add(max_iou_idx.item())
            else:
                fp[i] = 1

        tp_cumsum = torch.cumsum(tp, dim=0)
        fp_cumsum = torch.cumsum(fp, dim=0)

        precision = tp_cumsum / (tp_cumsum + fp_cumsum + 1e-6)
        recall = tp_cumsum / len(gt_boxes)

        precision = precision.numpy()
        recall = recall.numpy()
        recall_thresholds = np.linspace(0, 1, 101)
        precisions = np.zeros_like(recall_thresholds)
        for i, t in enumerate(recall_thresholds):
            precisions[i] = np.max(precision[recall >= t]) if np.any(recall >= t) else 0
        
        average_precision = np.mean(precisions)
        average_precisions.append(average_precision)

    mAP = sum(average_precisions) / len(average_precisions) if average_precisions else 0
    return mAP

def filter_json_by_map(input_file, output_file, map_threshold=0.4):
    with open(input_file, 'r', encoding='utf-8') as file:
        data = json.load(file)
    
    filtered_data = []

    for item in data:
        result_data = [
            [res['class_id'], res['accuracy'], res['coordinates']['x_min'], res['coordinates']['y_min'], res['coordinates']['x_max'], res['coordinates']['y_max']]
            for res in item['results']
        ]
        answer_data = [
            (ans['class_id'], [ans['coordinates']['x_min'], ans['coordinates']['y_min'], ans['coordinates']['x_max'], ans['coordinates']['y_max']])
            for ans in item['answer']
        ]
        
        mAP_value = mAP_calculate(answer_data, result_data)
        if mAP_value >= map_threshold:
            filtered_data.append(item)
    
    with open(output_file, 'w', encoding='utf-8') as file:
        json.dump(filtered_data, file, ensure_ascii=False, indent=4)

# Define paths
input_file_path = "/ai-video-converter/src/custom.json"
output_file_path = "/ai-video-converter/src/custom_1.json"

# Filter JSON data by mAP and save the result
filter_json_by_map(input_file_path, output_file_path)

print(f"Filtered data saved to {output_file_path}")
```

```python
import json
import os

def load_json(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)
    return data

def save_json(data, file_path):
    with open(file_path, 'w', encoding='utf-8') as file:
        json.dump(data, file, ensure_ascii=False, indent=4)

def fix_input_paths(json_data):
    for item in json_data:
        input_path = item['input_path']
        
        # Split the input path into parts
        parts = input_path.split('/')
        
        # Check if the last part is a folder with the same name as the file (without extension)
        file_name = parts[-1]
        folder_name = parts[-2]
        if os.path.splitext(file_name)[0] == folder_name:
            # Reconstruct the path without the redundant folder
            item['input_path'] = '/'.join(parts[:-2] + [file_name])
    
    return json_data

# Define paths
input_file_path = "/ai-video-converter/src/custom_3.json"
output_file_path = "/ai-video-converter/src/custom_4.json"

# Load JSON data
data = load_json(input_file_path)

# Fix input paths
fixed_data = fix_input_paths(data)

# Save the modified JSON data
save_json(fixed_data, output_file_path)

print(f"Modified data saved to {output_file_path}")
```

```python
import json
import os
import shutil
import random

def load_json(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)
    return data

def copy_random_images(json_data, source_base_path, target_base_path, num_images):
    # Ensure the target base path exists
    os.makedirs(target_base_path, exist_ok=True)
    
    # Filter entries to only include those from L02
    l02_entries = [entry for entry in json_data if '/L07/' in entry['input_path']]
    
    # Check if there are enough entries to sample from
    if len(l02_entries) < num_images:
        raise ValueError(f"Not enough entries to sample {num_images} images. Only found {len(l02_entries)} entries.")
    
    # Randomly select 200 entries from the filtered list
    selected_entries = random.sample(l02_entries, num_images)
    
    for entry in selected_entries:
        input_path = entry['input_path']
        
        # Determine the relative path of the image file to create necessary subdirectories
        relative_path = os.path.relpath(input_path, source_base_path)
        target_path = os.path.join(target_base_path, relative_path)
        target_dir = os.path.dirname(target_path)
        
        # Ensure the target directory exists
        os.makedirs(target_dir, exist_ok=True)
        
        # Copy the file to the target directory
        if os.path.exists(input_path):
            shutil.copy2(input_path, target_path)
            print(f"Copied {input_path} to {target_path}")
        else:
            print(f"File not found: {input_path}")

# Define paths
json_file_path = "/ai-video-converter/src/custom_3.json"
source_base_path = "/ai-video-converter/datasets/sample_Data/image/L07"
target_base_path = "/ai-video-converter/datasets/custom_data/image/L07"
num_images_to_copy = 500

# Load JSON data
data = load_json(json_file_path)

# Copy random images
copy_random_images(data, source_base_path, target_base_path, num_images_to_copy)
```

```python
import os
import shutil

# Define base paths
image_base_path = "/ai-video-converter/datasets/custom_data/image"
label_source_base_path = "/ai-video-converter/datasets/sample_Data/label"
label_target_base_path = "/ai-video-converter/datasets/custom_data/label"

def ensure_dir_exists(path):
    if not os.path.exists(path):
        os.makedirs(path)

def copy_corresponding_json(image_file_path):
    # Construct the relative path from the image base path
    relative_path = os.path.relpath(image_file_path, image_base_path)
    
    # Construct the corresponding label file path
    label_file_path = os.path.join(label_source_base_path, relative_path)
    label_file_path = os.path.splitext(label_file_path)[0] + ".json"
    
    # Check if the corresponding label file exists
    if os.path.exists(label_file_path):
        # Construct the target label file path
        target_label_file_path = os.path.join(label_target_base_path, relative_path)
        target_label_file_path = os.path.splitext(target_label_file_path)[0] + ".json"
        
        # Ensure the target directory exists
        ensure_dir_exists(os.path.dirname(target_label_file_path))
        
        # Copy the label file
        shutil.copy2(label_file_path, target_label_file_path)
        print(f"Copied {label_file_path} to {target_label_file_path}")
    else:
        print(f"Label file does not exist: {label_file_path}")

def process_images():
    # Walk through the image directory
    for root, dirs, files in os.walk(image_base_path):
        for file in files:
            if file.lower().endswith(('.jpg', '.jpeg', '.png')):
                image_file_path = os.path.join(root, file)
                copy_corresponding_json(image_file_path)

# Run the process
process_images()
```

```python
import os
import json

def load_json(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return json.load(file)

def count_classes(label_base_path):
    human_count = 0
    car_count = 0

    # Walk through the label directory
    for root, dirs, files in os.walk(label_base_path):
        for file in files:
            if file.lower().endswith('.json'):
                json_file_path = os.path.join(root, file)
                data = load_json(json_file_path)
                
                annotations = data.get("Learning_Data_Info.", {}).get("Annotations", [])
                for annotation in annotations:
                    class_id = annotation.get("Class_ID", "")
                    if class_id == "Human":
                        human_count += 1
                    elif class_id == "Car":
                        car_count += 1

    return human_count, car_count

def main():
    label_base_path = "/ai-video-converter/datasets/sample_Data/label/"
    
    human_count, car_count = count_classes(label_base_path)
    
    print(f"Total Human count: {human_count}")
    print(f"Total Car count: {car_count}")

if __name__ == "__main__":
    main()
```

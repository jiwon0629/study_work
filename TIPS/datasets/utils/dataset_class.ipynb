```python
import matplotlib.pyplot as plt

# Data for the counts
l_groups = ['L01', 'L02', 'L03', 'L04', 'L05', 'L06', 'L07', 'L08', 'L10']
car_counts = [1022, 4096, 1113, 268, 1595, 3130, 998, 1995, 0]
human_counts = [5250, 386, 417, 473, 2756, 78, 37, 94, 5171]

# Set up the bar width
bar_width = 0.35
index = range(len(l_groups))

# Plotting the data
fig, ax = plt.subplots()

bar1 = plt.bar(index, car_counts, bar_width, label='Car')
bar2 = plt.bar([i + bar_width for i in index], human_counts, bar_width, label='Human')

# Adding labels and title
plt.xlabel('L Groups')
plt.ylabel('Count')
plt.title('Counts of JSON files containing \'Car\' and \'Human\' class IDs by L Group')
plt.xticks([i + bar_width / 2 for i in index], l_groups)
plt.legend()

# Adding the count labels on the bars
for bar in bar1:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2.0, yval, int(yval), va='bottom')  # va: vertical alignment

for bar in bar2:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2.0, yval, int(yval), va='bottom')  # va: vertical alignment

# Show the plot
plt.show()
```
